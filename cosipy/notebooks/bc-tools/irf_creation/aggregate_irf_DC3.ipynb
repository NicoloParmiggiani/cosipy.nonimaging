```python
# Open the simulated irf to aggregate the crystals in the 6 final modules
irf_path = "/data/models/irf_run17.h5"
drm_list = DRMList.open(irf_path)
```

```python

# Function to recursively print the structure of the HDF5 file
def print_structure(name, obj):
    print(f"{name}: {type(obj)}")
    if isinstance(obj, h5py.Group):  # If it's a group, recursively explore it
        for key in obj:
            print_structure(key, obj[key])

# Open the HDF5 file in read-only mode ('r')
with h5py.File(irf_path, 'r') as h5_file:
    # Start from the root group to print the entire structure
    print_structure('/', h5_file)
```

```python

# Apri il file originale in modalità lettura
with h5py.File(irf_path, 'r') as h5_file:
    
    attributes = dict(h5_file.attrs)

    
    # Accedi ai dataset esistenti
    BGO_Z0_0 = h5_file['BGO_Coinc_bottom1']
    BGO_Z0_1 = h5_file['BGO_Coinc_bottom2']
    BGO_Z0_2 = h5_file['BGO_Coinc_bottom3']
    BGO_Z0_3 = h5_file['BGO_Coinc_bottom4']
    BGO_Z0_4 = h5_file['BGO_Coinc_bottom5']
    
    BGO_Z1_0 = h5_file['BGO_Coinc_bottom10']
    BGO_Z1_1 = h5_file['BGO_Coinc_bottom9']
    BGO_Z1_2 = h5_file['BGO_Coinc_bottom8']
    BGO_Z1_3 = h5_file['BGO_Coinc_bottom7']
    BGO_Z1_4 = h5_file['BGO_Coinc_bottom6']

    # Leggi gli attributi
    attributes_dataset = {}
    for attr_name, attr_value in h5_file['BGO_Coinc_bottom1'].attrs.items():
        attributes_dataset[attr_name] = attr_value

    BGO_X1_0 = h5_file['BGO_Coinc_sideX_3']
    BGO_X1_1 = h5_file['BGO_Coinc_sideX_2']
    BGO_X1_2 = h5_file['BGO_Coinc_sideX_1']

    BGO_X0_0 = h5_file['BGO_Coinc_sideX_neg_1']
    BGO_X0_1 = h5_file['BGO_Coinc_sideX_neg_2']
    BGO_X0_2 = h5_file['BGO_Coinc_sideX_neg_3']

    BGO_Y1_0 = h5_file['BGO_Coinc_sideY_1']
    BGO_Y1_1 = h5_file['BGO_Coinc_sideY_2']
    BGO_Y1_2 = h5_file['BGO_Coinc_sideY_3']

    BGO_Y0_0 = h5_file['BGO_Coinc_sideY_neg_3']
    BGO_Y0_1 = h5_file['BGO_Coinc_sideY_neg_2']
    BGO_Y0_2 = h5_file['BGO_Coinc_sideY_neg_1']

    
    
    # Controlla la forma dei dataset per vedere se sono compatibili

    # Somma i due dataset
    BGO_X1 = BGO_X1_0[:] + BGO_X1_1[:] + BGO_X1_2[:]
    BGO_X0 = BGO_X0_0[:] + BGO_X0_1[:] + BGO_X0_2[:]

    BGO_Y1 = BGO_Y1_0[:] + BGO_Y1_1[:] + BGO_Y1_2[:]+ BGO_Z1_4[:]
    BGO_Y0 = BGO_Y0_0[:] + BGO_Y0_1[:] + BGO_Y0_2[:] + BGO_Z0_4[:]

    BGO_Z1 = BGO_Z1_0[:]+BGO_Z1_1[:]+BGO_Z1_2[:]+BGO_Z1_3[:]
    BGO_Z0 = BGO_Z0_0[:]+BGO_Z0_1[:]+BGO_Z0_2[:]+BGO_Z0_3[:]
    
    # Crea un nuovo file HDF5 per salvare i dati
    with h5py.File(irf_path.replace("irf","irf_summed"), 'w') as new_h5_file:
        # Copia i dataset non sommati nel nuovo file
        #h5_file.copy('BGO_Coinc_sideX', BGO_Coinc_sideX)

        for key, value in attributes.items():
            new_h5_file.attrs[key] = value

        dataset_BGO_Z1 = new_h5_file.create_dataset('BGO_Z1', data=BGO_Z1)
        # Aggiungi gli attributi
        for attr_name, attr_value in attributes_dataset.items():
            dataset_BGO_Z1.attrs[attr_name] = attr_value
            
        dataset_BGO_Z0 = new_h5_file.create_dataset('BGO_Z0', data=BGO_Z0)
        # Aggiungi gli attributi
        for attr_name, attr_value in attributes_dataset.items():
            dataset_BGO_Z0.attrs[attr_name] = attr_value

        # Salva il risultato della somma nel nuovo file
        dataset_BGO_X1 = new_h5_file.create_dataset('BGO_X1', data=BGO_X1)
        # Aggiungi gli attributi
        for attr_name, attr_value in attributes_dataset.items():
            dataset_BGO_X1.attrs[attr_name] = attr_value
        
        dataset_BGO_X0 = new_h5_file.create_dataset('BGO_X0', data=BGO_X0)
        # Aggiungi gli attributi
        for attr_name, attr_value in attributes_dataset.items():
            dataset_BGO_X0.attrs[attr_name] = attr_value
            
        dataset_BGO_Y1 = new_h5_file.create_dataset('BGO_Y1', data=BGO_Y1)
        # Aggiungi gli attributi
        for attr_name, attr_value in attributes_dataset.items():
            dataset_BGO_Y1.attrs[attr_name] = attr_value
            
        dataset_BGO_Y0 = new_h5_file.create_dataset('BGO_Y0', data=BGO_Y0)
        # Aggiungi gli attributi
        for attr_name, attr_value in attributes_dataset.items():
            dataset_BGO_Y0.attrs[attr_name] = attr_value
```
