```python
# Open the simulated irf to aggregate the crystals in the 6 final modules
irf_path = "/data/models/irf_run20.h5"
drm_list = DRMList.open(irf_path)
```

```python

# Function to recursively print the structure of the HDF5 file
def print_structure(name, obj):
    print(f"{name}: {type(obj)}")
    if isinstance(obj, h5py.Group):  # If it's a group, recursively explore it
        for key in obj:
            print_structure(key, obj[key])

# Open the HDF5 file in read-only mode ('r')
with h5py.File(irf_path, 'r') as h5_file:
    # Start from the root group to print the entire structure
    print_structure('/', h5_file)
```

```python

with h5py.File(irf_path, 'r') as h5_file:
    
    attributes = dict(h5_file.attrs)

    ACS_Z0_0 = h5_file['ACS_Z0_0']
    ACS_Z0_1 = h5_file['ACS_Z0_1']
    ACS_Z0_2 = h5_file['ACS_Z0_2']
    ACS_Z0_3 = h5_file['ACS_Z0_3']
    ACS_Z0_4 = h5_file['ACS_Z0_4']
    
    ACS_Z1_0 = h5_file['ACS_Z1_0']
    ACS_Z1_1 = h5_file['ACS_Z1_1']
    ACS_Z1_2 = h5_file['ACS_Z1_2']
    ACS_Z1_3 = h5_file['ACS_Z1_3']
    ACS_Z1_4 = h5_file['ACS_Z1_4']

    attributes_dataset = {}
    for attr_name, attr_value in h5_file['ACS_Z0_0'].attrs.items():
        attributes_dataset[attr_name] = attr_value

    ACS_X1_0 = h5_file['ACS_X1_0']
    ACS_X1_1 = h5_file['ACS_X1_1']
    ACS_X1_2 = h5_file['ACS_X1_2']

    ACS_X0_0 = h5_file['ACS_X0_0']
    ACS_X0_1 = h5_file['ACS_X0_1']
    ACS_X0_2 = h5_file['ACS_X0_2']

    ACS_Y1_0 = h5_file['ACS_Y1_0']
    ACS_Y1_1 = h5_file['ACS_Y1_1']
    ACS_Y1_2 = h5_file['ACS_Y1_2']

    ACS_Y0_0 = h5_file['ACS_Y0_0']
    ACS_Y0_1 = h5_file['ACS_Y0_1']
    ACS_Y0_2 = h5_file['ACS_Y0_2']

    ACS_X1 = ACS_X1_0[:] + ACS_X1_1[:] + ACS_X1_2[:]
    ACS_X0 = ACS_X0_0[:] + ACS_X0_1[:] + ACS_X0_2[:]

    ACS_Y1 = ACS_Y1_0[:] + ACS_Y1_1[:] + ACS_Y1_2[:]+ ACS_Z1_0[:]
    ACS_Y0 = ACS_Y0_0[:] + ACS_Y0_1[:] + ACS_Y0_2[:] + ACS_Z0_4[:]

    ACS_Z1 = ACS_Z1_1[:]+ACS_Z1_2[:]+ACS_Z1_3[:]+ACS_Z1_4[:]
    ACS_Z0 = ACS_Z0_0[:]+ACS_Z0_1[:]+ACS_Z0_2[:]+ACS_Z0_3[:]
    
    
    with h5py.File(irf_path.replace("irf","irf_summed"), 'w') as new_h5_file:
       
        #h5_file.copy('ACS_Coinc_sideX', ACS_Coinc_sideX)

        for key, value in attributes.items():
            new_h5_file.attrs[key] = value

        dataset_ACS_Z1 = new_h5_file.create_dataset('ACS_Z1', data=ACS_Z1)
        
        for attr_name, attr_value in attributes_dataset.items():
            dataset_ACS_Z1.attrs[attr_name] = attr_value
            
        dataset_ACS_Z0 = new_h5_file.create_dataset('ACS_Z0', data=ACS_Z0)
       
        for attr_name, attr_value in attributes_dataset.items():
            dataset_ACS_Z0.attrs[attr_name] = attr_value


        dataset_ACS_X1 = new_h5_file.create_dataset('ACS_X1', data=ACS_X1)
  
        for attr_name, attr_value in attributes_dataset.items():
            dataset_ACS_X1.attrs[attr_name] = attr_value
        
        dataset_ACS_X0 = new_h5_file.create_dataset('ACS_X0', data=ACS_X0)
        
        for attr_name, attr_value in attributes_dataset.items():
            dataset_ACS_X0.attrs[attr_name] = attr_value
            
        dataset_ACS_Y1 = new_h5_file.create_dataset('ACS_Y1', data=ACS_Y1)
        
        for attr_name, attr_value in attributes_dataset.items():
            dataset_ACS_Y1.attrs[attr_name] = attr_value
            
        dataset_ACS_Y0 = new_h5_file.create_dataset('ACS_Y0', data=ACS_Y0)
  
        for attr_name, attr_value in attributes_dataset.items():
            dataset_ACS_Y0.attrs[attr_name] = attr_value
```
