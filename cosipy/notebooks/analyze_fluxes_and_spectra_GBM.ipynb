```python
import pandas as pd
import numpy as np
file = "sgrb_fluxes_bis.csv"
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv(file)


import ast
df["model parameters"] = df["model parameters"].apply(ast.literal_eval)

print(df)
```

```python


df_filtered = df[df["t90 (s)"] < 2]

energy_flux_filtered = df_filtered["photon flux (ph/cm2/s for 10-10000 keV)"]

bins = np.logspace(np.log10(energy_flux_filtered.min()), np.log10(energy_flux_filtered.max()), num=30)


plt.hist(energy_flux_filtered, bins=bins, edgecolor='black', alpha=0.7)
plt.xscale("log")

X = 1 
Y = 30 

plt.axvline(X, color='red', linestyle='dashed', linewidth=1.5, label=f'{X} ph/cm2/s')
plt.axvline(Y, color='blue', linestyle='dotted', linewidth=1.5, label=f'{Y} ph/cm2/s')


plt.xlabel('Energy Flux (ph/cm²/s)',fontsize=10)
plt.ylabel('Frequency',fontsize=10)
plt.title('')
plt.tick_params(axis='both', labelsize=10) 
plt.legend(fontsize=10)

plt.show()
```
