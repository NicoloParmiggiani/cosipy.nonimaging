```python
import numpy as np
import matplotlib.pyplot as plt
import sys,os
from pathlib import Path

# Configure here the path to the background data
data_dir_path = "/data/background/dc4"
```

```python
def calculate_hist_lc(file,bin_size,g_tmin,g_tmax):
    
    data_loaded = np.load(file)

    bgo_z1 = data_loaded['z1']
    bgo_z0 = data_loaded['z0']
    bgo_x1 = data_loaded['x1']
    bgo_x0 = data_loaded['x0']
    bgo_y1 = data_loaded['y1']
    bgo_y0 = data_loaded['y0']

    data_array = [bgo_z1,bgo_z0,bgo_x1,bgo_x0,bgo_y1,bgo_y0]
    
    print("len data"+str(len(data_loaded)))

    bin_edges = np.arange(g_tmin, g_tmax + bin_size, bin_size)
    
    ligth_curve = np.empty((len(bin_edges-2),6,2))
        
    for j in range(0,len(data_array)):
    
        frequenze, _ = np.histogram(data_array[j], bins=bin_edges)
        
        for i in range(len(frequenze)):

            ligth_curve[i][j][0] = bin_edges[i]
            ligth_curve[i][j][1] = frequenze[i]

    return ligth_curve
    
```

```python
def get_min_max(global_min,global_max,filename):

    data_loaded = np.load(filename)
    
    bgo_z1 = data_loaded['z1']
    bgo_z0 = data_loaded['z0']
    bgo_x1 = data_loaded['x1']
    bgo_x0 = data_loaded['x0']
    bgo_y1 = data_loaded['y1']
    bgo_y0 = data_loaded['y0']
    
    print("len data"+str(len(data_loaded)))
    
    all_counts = np.concatenate([bgo_z1,bgo_z0,bgo_x1,bgo_x0,bgo_y1,bgo_y0])
    
    local_min = min(all_counts)
    local_max = max(all_counts)

    if local_min<global_min:
        global_min=local_min
    if local_max>global_max:
        global_max = local_max

    return global_min,global_max
    
```

```python
# find a window of 500 s with a mean background
time_series= total_counts[:,2,0][start_orbit:stop_orbit]
window_size = 10000
target_mean = mean_counts_orbit

best_start = 0
min_diff = float('inf')

for i in range(len(time_series) - window_size + 1):
    window = time_series[i:i+window_size]
    mean = np.mean(window)
    diff = abs(mean - target_mean)
    
    if diff < min_diff:
        min_diff = diff
        best_start = i

best_window = time_series[best_start:best_start+window_size]
```
